```python
from pyspark.sql import SparkSession
from data_profiling import ProfileReport

spark = SparkSession.builder \
    .appName("profiling") \
    .config("spark.sql.ansi.enabled", "false") \
    .config("spark.driver.memory", "16g") \
    .getOrCreate()

parquet_name = "4c_eea_co2_emissions_from_passenger_cars-001.parquet"
path = f"../data/cleaned/{parquet_name}"

df = spark.read.parquet(path)

profile = ProfileReport(df, title=f"Data Profile: {parquet_name}")
profile.to_file("reports/4c_eea_co2_emissions_from_passenger_cars-001.html")
```
